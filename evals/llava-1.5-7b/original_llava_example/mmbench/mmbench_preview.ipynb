```python
answer_char = "A"
shift = 2
options = ["A", "B", "C", "D"]
# find the index of the answer character
print(options)
n = len(options)
def shift_list(lst, n):
    return lst[n:] + lst[:n]

shifted_options = shift_list(options, shift)
print(shifted_options)
def shift_answer(n):
    answer_index = options.index(answer_char)
    shifted_index = (answer_index - n) % len(options)
    return options[shifted_index]

print(shift_answer(shift))

```

```python
# Load the dataset
import os
questions = pd.read_table(os.path.expanduser("mmbench_dev_20230712.tsv"))
print(f"Dataset shape: {questions.shape}")
print(f"Dataset size: {len(questions)} rows")

# Get the whole dataframe (equivalent to get_chunk(questions, 1, 0))
questions_chunked = questions
print(f"\nquestions_chunked shape: {questions_chunked.shape}")
print(f"Size: {len(questions_chunked)} rows")
```

```python
import math

def split_list(lst, n):
    """Split a list into n (roughly) equal-sized chunks"""
    chunk_size = math.ceil(len(lst) / n)  # integer division
    return [lst[i : i + chunk_size] for i in range(0, len(lst), chunk_size)]


def get_chunk(lst, n, k):
    chunks = split_list(lst, n)
    return chunks[k]
```

```python
questions = get_chunk(questions, 1, 0)
print(f"\nquestions shape: {questions.shape}")
```
